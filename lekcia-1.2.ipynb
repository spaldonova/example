```python
# Vytvor variable score s hodnotou medzi 0 a 100. Použi control flows na priradenie známky (A: 90-100, B: 80-89, C: 70-79, D: 60-69, F: pod 60).

score = 85
if score >=90:
    grade = 'A'
elif score >=80:
    grade = 'B'
elif score >=70: 
    grade = 'C' 
elif score >=60:
    grade = 'D' 
else:
    grade = 'F' 
    
print('Tvoja znamka je:', grade)
```

```python
# Vytvor list boolean náhodných True alebo False hodnôt. Vypočítaj a printni percento True prvkov.
nahodny_zoznam = [True, True, False, False, True, True, False, True ]
true_pocet = nahodny_zoznam.count(True)
percento = (true_pocet /len(nahodny_zoznam)) * 100
print('Percento je:', percento)
```

```python
# Vytvor list floating-point čísel. Uprav list tak, aby každé číslo bolo zaokrúhlené na 2 desatinné miesta.
zoznam_floating_point = [4.7778, 5.6685, 9.3, 56.888997]
zaokruhlene_cisla = [round(num, 2) for num in zoznam_floating_point]

print('Zaokruhlene cisla na 2 desatinne miesta:', zaokruhlene_cisla)
```

```python
# Vytvor variable day_idx s hodnotou medzi 1 a 365. Použi control flows na určenie toho, ktorý deň v týždni to je. (Hint: Zváž ako vieš využiť modulo operátor pri hľadaní vzoru.)
day_idx = 33
days = ["Pondelok", "Utorok", "Streda", "Štvrtok", "Piatok", "Sobota", "Nedeľa"]
day_of_week = days[(day_idx -1 ) % 7]
print(f"Deň číslo {day_idx} v roku je {day_of_week}.")
```

```python
# Vytvor list stringov reprezentujúcich názvy súborov (aj s koncovkami ako '.pdf', '.txt', '.jpg', '.docx' atď.). Použi loops na odfiltrovanie stringov, ktoré nereprezentujú textové súbory a printni výsledok.
files = ["dokument.pdf", "obrazok.jpg", "poznamky.txt", "prezentacia.pptx", "text.docx", "hudba.mp3"]
text_files = [] 
 
for file in files:
    if file.endswith(('.pdf', '.txt', '.docx')):
        text_files.append(file)
        
        
print('Textove subory:', text_files)
```

```python
# Vytvor list obsahujúci dictionaries, kde každý dictionary reprezentuje osobu s menom a vekom. Vypočítaj a printni priemerný vek.
zoznam_osob = [{'meno': 'Iva', 'vek': 44}, {'meno': 'Jana', 'vek': 33}, {'meno': 'Jozef', 'vek': 77}]
celkovy_vek = 0
for osoba in zoznam_osob:
    celkovy_vek += osoba['vek']
    priemerny_vek = celkovy_vek/len(osoba)
print(priemerny_vek)
```
