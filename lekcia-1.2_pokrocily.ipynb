```python
#uloha 4 Matrix v Python je list listov obsahujúci rovnaký počet položiek rovnakého typu (napr. čísel). Vytvor niekoľko vzorových listov listov a použi control flows na to, aby si zistila, či sú to matrixy (matice).

m1 = [[1,2,3],
      [2,5,6],
      [7,8,9]]
m2 = [[1.5,2.5,3.3],
      [2.8,3.5,4.4]]
m3 = [['s',2,6],
      ['m',8,1.2]]
m4 = [[1,5,6],
      [1,8]]
#matrix = [m1, m2, m3, m4]

def is_matrix(matrix):
    
# overime dlzku matic
    first_row_length = len(matrix[0])
    for row in matrix:
        if len(row) != first_row_length:
            return False
    
# overime ci vsetky prvky su rovnakeho typu
    first_type = type(matrix[0][0])
    for row in matrix:
        for item in row:
            if type(item) != first_type:
                return False
        
    return True

matrix = [m1, m2, m3, m4]

for item, m in enumerate(matrix, start=1):
    if is_matrix(m):
        print(f"m{item} JE matica ✅")
    else:
        print(f"m{item} NIE JE matica ❌")
```

```python
# uloha 7 - Vytvor list integerov medzi 1 a 50. Zisti, koľko z nich sú prvočísla.

def je_prvocislo(n):
    if n<=1:
        return False
    for i in range(2,int(n ** 0.5) + 1): # kontrolujeme len po druhu odmocninu
        if n % i ==0:
            return False
    return True



cisla = list(range(1,51))

prvocisla = [n for n in cisla if je_prvocislo]
print("Prvočísla:", prvocisla)
print("Počet prvočísel:", len(prvocisla))
```

```python
# Funkcia na určenie, či je číslo prvočíslo
def je_prvocislo(n):
    if n <= 1:
        return False
    for i in range(2, int(n ** 0.5) + 1):  # kontrolujeme len po druhú odmocninu
        if n % i == 0:
            return False
    return True

# List čísel od 1 do 50
cisla = list(range(1, 51))

# Vyberieme len prvočísla (pozor na volanie funkcie!)
prvocisla = [n for n in cisla if je_prvocislo(n)]

# Výpis výsledkov
print("Prvočísla:", prvocisla)
print("Počet prvočísel:", len(prvocisla))
```

```python
# uloha 9 - Vytvor dva listy so 7 náhodnými číslami medzi 0 a 9. Vytvor dictionary obsahujúci ich union, intersection a symmetric difference.

list_1 = [1,5,4,8,6,7,3]
list_2 = [2,3,4,6,8,9,1]

set_1 = set(list_1)
set_2 = set(list_2)

vysledky = {'union': set_1 | set_2,  # zjednotenie  zoberie vsetko
            'intersection': set_1 & set_2,   # prienik  zoberie iba duplikaty
            'symmetric_difference': set_1 ^ set_2}   # symetricka diferencia, zoberie vsetko okrem duplikatov
print(vysledky)
```
