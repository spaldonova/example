```python
# Počiatočné hodnoty
pociatocny_vklad = 1000
mesacny_vklad = 100
rocna_urokova_sadzba = 0.021
pocet_mesiacov = 24

# Mesačná úroková sadzba
mesacna_sadzba = rocna_urokova_sadzba / 12

# Výpočet nasporenej sumy
zostatok = pociatocny_vklad

for mesiac in range(pocet_mesiacov):
    zostatok = zostatok * (1 + mesacna_sadzba) + mesacny_vklad

# Výstup
print("Nasporená suma po 2 rokoch:", round(zostatok, 2), "USD")
```

```python
seconds = 3749

# 1 hodina = 3600 sekúnd
# 1 minúta = 60 sekúnd

hours = seconds // 3600              # celé hodiny
remaining_seconds = seconds % 3600   # zvyšné sekundy po odrátaní hodín
minutes = remaining_seconds // 60    # celé minúty

print("Hodiny:", hours)
print("Minúty:", minutes)

```

```python
# 1. január 2025 = streda
zaciatocny_den = 3   # 1=pondelok, 2=utorok, 3=streda, ..., 7=nedeľa
den_roka = 175

# Počet dní, ktoré uplynuli od 1. januára
presun = den_roka - 1

# Výpočet dňa v týždni
den_v_tyzdni = (zaciatocny_den + presun) % 7

# Ak výsledok je 0, znamená to nedeľu
if den_v_tyzdni == 0:
    den_v_tyzdni = 7

# Mapovanie čísla na názov dňa
dni = {
    1: "Pondelok",
    2: "Utorok",
    3: "Streda",
    4: "Štvrtok",
    5: "Piatok",
    6: "Sobota",
    7: "Nedeľa"
}

print("175. deň roku 2025 padol na:", dni[den_v_tyzdni])
```

```python
# Počet šálok múky pre 5 ľudí
cups_for_5 = 4

# Počet ľudí, pre ktorých chceme recept
people = 7

# --- Integer verzia ---
cups_per_person_int = cups_for_5 // 5      # celé delenie
cups_needed_int = cups_per_person_int * people

# --- Float verzia ---
cups_per_person_float = cups_for_5 / 5     # desatinné delenie
cups_needed_float = cups_per_person_float * people

print("Integer verzia: ", cups_needed_int, "šálok múky")
print("Float verzia:   ", cups_needed_float, "šálok múky")
```

```python
recipe_people = 5
actual_people = 7

## výpočet ako integer
cups_int = 4
scaled_int = cups_int / recipe_people * actual_people ## vypočíta množstvo na osobu a potom násobí novým počtom ľudí

## výpočet ako float
cups_float = 4.0
scaled_float = cups_float / recipe_people * actual_people

print(scaled_int, scaled_float) 
```
