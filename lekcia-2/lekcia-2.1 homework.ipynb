```python
# ULOHA 13 Napíš funkciu fibonacci(), ktorá berie integer n a vráti n-té Fibonacci číslo.
# number_start = [0]
def fibonacci(n):
    if n <= 0:
        return 'Zadaj kladne cislo'
    elif n == 1:
        return 0
    elif n == 2:
        return 1
    else:
        return fibonacci(n - 1)+fibonacci(n-2)
    
print(fibonacci(11))
    
```

```python
# uloha 4 - Vytvor funkciu leap_year_check(), ktorá berie rok a vráti True, ak je rok prestupný.
# Rok je prestupný, ak:
# je deliteľný 4,
# ale nie deliteľný 100,
# okrem prípadu, že je zároveň deliteľný 400.

def leap_year_check(year):
    if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
        return True
    else:
        return False
    
print(leap_year_check(2020))
```

```python
# uloha 5 - Napíš funkciu greet_everyone(), ktorá berie list mien a vypíše "Hello, [meno]!" pre všetky mená v liste. Počet mien nie je vopred špecifikovaný.
list_mien = ['Janka', 'Maria', 'Ivana']
def greet_everyone(meno):
    for meno in list_mien:
        return 'Hello', (meno) # chyba return je vnutri cyklu, skonci po prvom pozdrave
    
print(greet_everyone(list_mien))
    
```

```python
# uloha 7 - Napíš funkciu prime_check(), ktorá berie integer a vráti True, ak je číslo prvočíslo, inak False. (Tip: uisti sa, že ošetríš záporné čísla a iné edge cases.)

def prime_check(x):
    if x <= 1:
        return False
    for i in range(2, int(x ** 0.5) + 1): # testuje sa len po odmocninu x, to staci na overenie prvocisel
        if x % i == 0:
            return False
    return True
        
              
print(prime_check(22))
```

```python
# uloha 8 - Vytvor funkciu word_count(), ktorá berie string s vetou a vráti dictionary s každým slovom ako key a jeho frekvenciou ako value.

def word_count(veta):
    veta = veta.lower()
    slova = veta.split()
    #x_ako_dict =x_ako_list.dict()
    freq_dict = {}
    for slovo in slova:
        if slovo in freq_dict:
            freq_dict[slovo] +=1
        else:
            freq_dict[slovo] = 1
    return freq_dict
   
   
veta = "Toto je test toto je len test"
print(word_count(veta))
    
```
