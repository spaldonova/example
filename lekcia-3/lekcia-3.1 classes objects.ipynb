```python
import random

class HadajCislo:
    def __init__(self):
        self.tajne_cislo = random.randint(1, 100)
        self.pokusy = 7

    def ziskaj_tip(self):
        """Získa vstup od hráčky a skontroluje, či je číslo platné."""
        while True:
            vstup = input(f"Zadaj svoj tip (zostáva {self.pokusy} pokusov): ")
            if vstup.isdigit():
                return int(vstup)
            print("❗ Prosím, zadaj celé číslo.")

    def ohodnot_tip(self, tip):
        """Vyhodnotí, ako blízko je tip k tajnému číslu."""
        rozdiel = abs(tip - self.tajne_cislo)
        if rozdiel == 0:
            print("🎉 Trafila si číslo!")
            return True
        elif rozdiel >= 20:
            print("❄️ Veľmi ďaleko!")
        elif 10 <= rozdiel <= 19:
            print("😕 Ďaleko!")
        elif 5 <= rozdiel <= 9:
            print("🙂 Blízko!")
        elif 1 <= rozdiel <= 4:
            print("🔥 Veľmi blízko!")
        return False

    def hraj(self):
        """Spustí hlavnú hru."""
        print("Vitaj v hre HÁDAJ ČÍSLO!")
        print("Počítač vybral číslo medzi 1 a 100. Máš 7 pokusov.\n")

        while self.pokusy > 0:
            tip = self.ziskaj_tip()

            if self.ohodnot_tip(tip):
                break  # trafila si číslo, koniec hry

            self.pokusy -= 1
            print()  # prázdny riadok pre prehľadnosť

            if self.pokusy == 0:
                print(f"😢 Žiaľ, prehrala si. Správne číslo bolo {self.tajne_cislo}.")

# Spustenie hry
hra = HadajCislo()
hra.hraj()
```
