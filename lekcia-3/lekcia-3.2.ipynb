```python
# uloha 1 - Vytvor základnú triedu Vehicle s atribútmi speed, fuel_capacity a fuel_consumption. Vytvor podradené triedy Car, Motorcycle a Truck, ktoré dedia z Vehicle. Každá podradená trieda by mala mať metódu travel_distance(hours), ktorá vypočíta, ako ďaleko dokáže cestovať.

class Vehicle:
    def __init__(self, speed, fuel_capacity, fuel_consumption):
        self.speed = speed
        self.fuel_capacity = fuel_capacity
        self.fuel_consumption = fuel_consumption
        
class Car(Vehicle):
    def __init__(self, speed, fuel_capacity, fuel_consumption):
        super().__init__(speed, fuel_capacity, fuel_consumption)
        
        
    def travel_distance(self, hours): 
        
        ideal_distance =  hours * self.speed  
        max_distance =  self.fuel_capacity/self.fuel_consumption * 100
        if ideal_distance > max_distance:
            return max_distance
        else:
            return ideal_distance
      
    
    
        
    
```

```python
red_car = Car(90,50,5)

print(red_car.travel_distance(30))
print(red_car.travel_distance(11.11))
```
