## 🎯 Zadanie

Definuj nasledujúce pojmy:

- *set* (množina) - neusporiadaná kolekcia jedinečných hodnôt.
To znamená:
neuchováva duplikáty,
nemá poradie prvkov (ako zoznam má indexy),
a podporuje matematické operácie s množinami – napr. zjednotenie, prienik, rozdiel.

- *subset* (podmnožina) - pojem z množín (set), ktorý označuje množinu, všetky prvky ktorej sa nachádzajú v inej množine.
Inými slovami:
Ak je množina A subsetom množiny B, všetky prvky A sú aj v B.
Môžeme ju overiť pomocou metódy issubset() alebo operátora <=.

- *union* (zjednotenie) - zjednotenie dvoch alebo viacerých množín.
Výsledná množina obsahuje všetky prvky zo všetkých množín, bez duplikátov.
Môžeme ju vytvoriť pomocou metódy union() alebo operátora |.

- *intersection* (prienik) - – množina všetkých prvkov, ktoré sa nachádzajú vo všetkých zúčastnených množinách.
Vráti len tie prvky, ktoré sú spoločné.
Môžeme ho vytvoriť pomocou metódy intersection() alebo operátora &.

- *complement* (komplement) - (doplnok) používa pri množinách a znamená všetky prvky, ktoré patria k jednej množine, ale nepatria do druhej.
Oficiálne v Pythone neexistuje priamo metóda complement(), ale doplnok sa dá získať pomocou rozdielu množín (-).
Inými slovami:
A complement B=A−B
Teda všetky prvky z A, ktoré nie sú v B.

- *difference* (rozdiel) - rozdiel množín – všetky prvky, ktoré sú v jednej množine, ale nie sú v druhej.
Je to vlastne to isté, čo doplnok (complement) z pohľadu jednej množiny.
Používame metódu difference() alebo operátor -.

- *symmetric difference* (symetrický rozdiel) – symetrický rozdiel dvoch množín.
Obsahuje všetky prvky, ktoré sú v jednej množine alebo v druhej, ale nie v oboch naraz.
Používame metódu symmetric_difference() alebo operátor ^.

Priprav si 3 otázky ohľadom hore spomenutých pojmov. Tieto otázky by mali byť špecifické a mali by sa sústrediť na nedostatky alebo medzery, na ktoré si narazila pri hľadaní definícii.
