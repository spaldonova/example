```python
# cesarova sifra

def ceasar_encypher(text: str, key: int = 3):
    text = input("Zadaj text, ktorý chceš zašifrovať: ")
    key = 3  # pevný posun o 3 písmená
    encrypted = []

    for char in text:
        if char.isalpha():  # len písmená
            base = ord('A') if char.isupper() else ord('a')
            encrypted += chr((ord(char) - base + key) % 26 + base)
        else:
            encrypted += char  # medzery, čísla a pod. sa nemenia
  


    print("Zašifrovaný text:" , encrypted)
    return encrypted
```
